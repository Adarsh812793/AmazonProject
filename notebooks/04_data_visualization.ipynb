**Presentation 4 Goal:** Demonstrate data visualization tools (`matplotlib` and `seaborn`) to communicate the exploratory data analysis findings clearly and effectively.

> **Scope Reminder:** This dataset contains **no sales or revenue data**. Visualizations represent catalog attributes, stated discount percentages, customer review counts, and star ratings.

All reusable plotting routines are defined in `src/amazon_project/visualization.py` and exported to `reports/figures/` for the presentation slides and project report.

## 1. Imports and Setup

We import Matplotlib, Seaborn, the processed data loader, and our custom visualization functions.

```python
import matplotlib.pyplot as plt

from amazon_project.loading import load_processed
from amazon_project.visualization import (
    generate_all_figures,
    plot_bivariate_relationships,
    plot_category_comparisons,
    plot_correlation_heatmap,
    plot_outlier_boxplots,
    plot_price_distributions,
    plot_rating_and_discount_distributions,
    save_figure,
    set_presentation_style,
)

# Apply standard clean presentation theme
set_presentation_style()
```

## 2. Load Processed Dataset

We load the verified clean dataset `data/processed/amazon_clean.csv`.

```python
df = load_processed()
print(f"Loaded {len(df):,} cleaned product records across {df.shape[1]} columns.")
df[
    [
        "product_name",
        "main_category",
        "discounted_price",
        "actual_price",
        "discount_percentage",
        "rating",
        "rating_count",
    ]
].head(3)
```

## 3. Product Price Distribution (Raw vs. Log10 Scale)

In Stage 4 EDA, we determined that `discounted_price` has a severe right skew (+4.30), with a median of ₹799 and a mean of ₹3,125 extending to ₹77,990.

- **Left Chart (Raw):** Highlights how extreme high-end electronics compress the bulk of products into the lower bins.
- **Right Chart (Log10):** Visualizes the distribution in log-space, revealing the underlying bell-shaped density of typical consumer purchases.

```python
fig1 = plot_price_distributions(df)
save_figure(fig1, "01_price_distributions.png")
plt.show()
```

## 4. Discounts and Customer Ratings Distributions

- **Discount Percentage:** Features a roughly symmetric, bell-shaped distribution centered around a 60.0% median, showing that aggressive catalog discounts (50% to 70%) are standard on Amazon India.
- **Customer Rating:** Highly left-skewed (-1.23) and tightly concentrated between 4.0 and 4.3 stars (median 4.10). Poorly rated items (<3.5 stars) are rare in search-promoted catalog listings.

```python
fig2 = plot_rating_and_discount_distributions(df)
save_figure(fig2, "02_discount_and_rating_distributions.png")
plt.show()
```

## 5. Outlier Detection with Tukey Fences

Using Tukey's 1.5 × IQR fences, Stage 4 identified:
- **114 price outliers** (> ₹4,911.5) — legitimate high-value items (smart TVs, DSLR gear, premium appliances).
- **124 rating count outliers** (> 38,391 reviews) — viral/staple accessories (charging cables, screen protectors).

The boxplots below illustrate these distributions on logarithmic scales.

```python
fig3 = plot_outlier_boxplots(df)
save_figure(fig3, "03_outlier_boxplots.png")
plt.show()
```

## 6. Category Comparison (Major Product Categories)

Comparing the 4 major categories with $\ge 5$ items:
- **Electronics:** Highest price variance (IQR ₹7,600) due to laptops and TVs.
- **OfficeProducts:** Very low discounts (median 5%) and low price points (median ₹178), yet highest customer satisfaction (median rating 4.3).
- **Home&Kitchen and Computers&Accessories:** Heavy discount ranges (median 60%+).

```python
fig4 = plot_category_comparisons(df)
save_figure(fig4, "04_category_comparisons.png")
plt.show()
```

## 7. Correlation Analysis

Spearman rank correlation heatmap showing monotonic relationships between all continuous variables.

```python
fig5 = plot_correlation_heatmap(df)
save_figure(fig5, "05_correlation_heatmap.png")
plt.show()
```

## 8. Bivariate Relationship Exploration

- **Discount % vs. Price ($r_s = -0.367$):** Moderate negative relationship — budget and mid-tier accessories often advertise 60–80% discounts, whereas expensive luxury electronics carry lower percent discounts.
- **Rating vs. Price ($r_s = +0.077$):** No meaningful relationship — higher price tags do not guarantee higher customer ratings.

```python
fig6 = plot_bivariate_relationships(df)
save_figure(fig6, "06_bivariate_relationships.png")
plt.show()
```
